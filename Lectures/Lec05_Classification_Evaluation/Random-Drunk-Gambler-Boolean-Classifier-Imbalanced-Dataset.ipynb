# Random Boolean Classifier on Imbalanced Dataset

A completely random classifier predicts 0 or 1 with a 50/50 chance. So, let's simulate this behaviour on an imbalanced dataset containing **1 positive** and **99 negative** class instances.

```python
import numpy as np
from sklearn.metrics import confusion_matrix

y_true = np.array([1] + [0] * 99)# Dataset: 1 positive, 99 negatives
y_pred = np.random.randint(0, 2, size=100)# Random classifier: 50% chance of predicting 1 or 0

tn, fp, fn, tp = confusion_matrix(y_true, y_pred, labels=[0, 1]).ravel()# sklearn order: [[TN, FP], [FN, TP]]

cm = np.array([# Preferred order: [[TP, FP], [FN, TN]]
    [tp, fp],
    [fn, tn]
])

print("Confusion matrix [[TP, FP], [FN, TN]]:")
print(cm)
```

### Many runs to obtain a better estimate

```python
import numpy as np
from sklearn.metrics import confusion_matrix

# Dataset: 1 positive, 99 negatives
y_true = np.array([1] + [0] * 99)

n_runs = 1000
cms = []

for _ in range(n_runs):
    y_pred = np.random.randint(0, 2, size=100)# Fair random classifier: 50% chance of predicting 0 or 1

    tn, fp, fn, tp = confusion_matrix(y_true, y_pred, labels=[0, 1]).ravel()# sklearn order: [[TN, FP], [FN, TP]]

    cm = [[tp, fp],
          [fn, tn]]
    cms.append(cm)

avg_cm = np.mean(cms, axis=0)
print("Average confusion matrix [TP, FP] / [FN, TN]:")
print(avg_cm)
```
