```python
from collections import Counter, defaultdict
docs = [
    """
    Ongchoi is a popular leafy green vegetable commonly used in
    Asian cuisine. Ongchoi has long hollow stems and tender green leaves.
    Many people cook ongchoi with garlic or add ongchoi to soups.
    """
]

context_size = 4
word_word_matrix = defaultdict(Counter)
for doc in docs:
    words = doc.split()
    for i, target in enumerate(words):
        # Context = words within ±contex_size
        start = max(0, i - context_size)
        end = min(len(words), i + context_size + 1)
        for j in range(start, end):
            if i != j:
                context_word = words[j]
                word_word_matrix[target][context_word] += 1

vocab = sorted(set(word for doc in docs for word in doc.split()))# get vocabs
print("          " + " ".join(f"{w:>7}" for w in vocab))
for word in vocab:
    row = [word_word_matrix[word][w] for w in vocab]
    print(f"{word:<8}" + " ".join(f"{x:>7}" for x in row))
```

**Dynamic Context: the entire doc**

```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer

docs = [
    "ongchoi is a popular leafy green vegetable",
    "ongchoi is commonly cooked with garlic",
    "people eat ongchoi in soups and stir fries"
]
count_model = CountVectorizer()# Document-Word matrix
X = count_model.fit_transform(docs)
Xc = X.T @ X# Document-level co-occurrence matrix
# Xc.setdiag(0)
words = count_model.get_feature_names_out()# Vocabulary
Xc_df = pd.DataFrame(Xc.toarray(), index=words, columns=words)
print(Xc_df)
```

**Document-Word Matrix**

```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer

docs = [
    "ongchoi is a popular leafy green vegetable",
    "ongchoi is commonly cooked with garlic",
    "people eat ongchoi in soups and stir fries"
]

count_model = CountVectorizer()
X = count_model.fit_transform(docs)
# X[X > 0] = 1 #Boolean space
words = count_model.get_feature_names_out()

# Print Document-Word matrix
X_df = pd.DataFrame(X.toarray(), index=[f"Doc {i+1}" for i in range(len(docs))], columns=words).transpose()
print(X_df.to_string())
```
